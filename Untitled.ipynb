```python
import pandas as pd
import matplotlib as plt
import seaborn as sns
import string
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
df = pd.read_csv(r"C:\heimdall\data\full_dataset.csv")
df.head()
print(df['category'].value_counts())
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def clean_text_fn(text):
    text = text.lower()  # lowercase
    text = ''.join([c for c in text if c not in string.punctuation])  # remove punctuation
    words = text.split()
    words = [lemmatizer.lemmatize(w) for w in words if w not in stop_words]  # remove stopwords & lemmatize
    return ' '.join(words)

df['clean_text'] = df['text'].apply(clean_text_fn)
```

```python
from sklearn.model_selection import train_test_split
X = df['clean_text']
y = df['category']
X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2,stratify=y,random_state=42)
print("Training samples:",len(X_train))
print("Test samples:",len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer(max_features=5000,stop_words='english')
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
import joblib
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
# model = MultinomialNB()
# model.fit(X_train_tfidf, y_train)
# y_pred = model.predict(X_test_tfidf)
models = {
    'Naive Bayes': MultinomialNB(),
    'Logistic Regression': LogisticRegression(),
    'Random Forest': RandomForestClassifier()
}

results = {}
for name, model in models.items():
    model.fit(X_train_tfidf, y_train)
    results[name] = model.score(X_test_tfidf, y_test)
    print(f"{name}: {results[name]:.4f}")  # Print as you go
# Save each model
for name, model in models.items():
    joblib.dump(model, f"{name.replace(' ', '_')}.pkl")

# Save the vectorizer too
joblib.dump(vectorizer, "tfidf_vectorizer.pkl")
```

```python
from sklearn.metrics import classification_report, accuracy_score, precision_score, recall_score, f1_score

results_detailed = {}

for name, model in models.items():
    model.fit(X_train_tfidf, y_train)
    y_pred = model.predict(X_test_tfidf)
    
    results_detailed[name] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, average='weighted'),
        'Recall': recall_score(y_test, y_pred, average='weighted'),
        'F1-Score': f1_score(y_test, y_pred, average='weighted')
    }
    
    print(f"\n{'='*50}")
    print(f"{name}")
    print('='*50)
    print(classification_report(y_test, y_pred))

# Create comparison DataFrame
comparison_df = pd.DataFrame(results_detailed).T
print("\n=== Overall Comparison ===")
print(comparison_df.round(4))
```

```python
import re
import joblib

def predict_email():
    """
    Prompts user to input an email, loads saved models & vectorizer,
    and prints predictions from each model.
    """
    # 1️⃣ Load vectorizer and models
    vectorizer = joblib.load("tfidf_vectorizer.pkl")
    models = {
        'Naive Bayes': joblib.load("Naive_Bayes.pkl"),
        'Logistic Regression': joblib.load("Logistic_Regression.pkl"),
        'Random Forest': joblib.load("Random_Forest.pkl")
    }

    # 2️⃣ Take user input
    email_text = input("Enter the email text:\n> ")

    # 3️⃣ Clean text
    email_clean = clean_text_fn(email_text)

    # 4️⃣ Convert to TF-IDF
    email_tfidf = vectorizer.transform([email_clean])

    # 5️⃣ Get predictions
    print("\n=== Model Predictions ===")
    for name, model in models.items():
        pred = model.predict(email_tfidf)[0]
        try:
            proba = model.predict_proba(email_tfidf)
            confidence = proba.max() * 100
            print(f"{name}: {pred} ({confidence:.2f}% confident)")
        except AttributeError:
            print(f"{name}: {pred}")

    print("\n✅ Prediction complete.")

choice = 1
while choice == 1:
    predict_email()
    choice = int(input("Enter choice:"))
```
